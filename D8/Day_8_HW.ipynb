### Word Embeddings with BM25 and Sentence Transformers

This section demonstrates how to generate embeddings for a specific word within a given sentence using two different approaches: BM25 (a statistical retrieval function) and Sentence Transformers (a deep learning approach for semantic similarity).

```python
# Install necessary libraries if you haven't already
!pip install -qqq gensim sentence-transformers nltk rank_bm25

import nltk
from nltk.tokenize import word_tokenize
from rank_bm25 import BM25Okapi # Use rank_bm25 for BM25 functionality
from sentence_transformers import SentenceTransformer
import torch
import numpy as np

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt')
except LookupError: # Changed from nltk.downloader.DownloadError
    nltk.download('punkt', quiet=True)

# Also download 'punkt_tab' for tokenization if not found
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError: # Changed from nltk.downloader.DownloadError
    nltk.download('punkt_tab', quiet=True)

print("Libraries imported and NLTK punkt tokenizer checked/downloaded.")
```

```python
sentence = "The quick brown fox jumps over the lazy dog."
word_to_find = "fox"

print(f"Sample Sentence: '{sentence}'")
print(f"Word to find: '{word_to_find}'")
```

#### 1. BM25 (Best Match 25) Embedding

BM25 is a ranking function used in information retrieval. While not strictly an 'embedding' in the vector space sense, it can provide a relevance score for terms within documents. Here, we'll calculate the BM25 score of our target word against the sentence.

```python
# Tokenize the sentence to form the corpus (list of tokenized documents)
tokenized_corpus = [word_tokenize(sentence.lower())]

# Create a BM25 model using BM25Okapi from rank_bm25
bm25_model = BM25Okapi(tokenized_corpus)

# Tokenize the query word
tokenized_query = word_tokenize(word_to_find.lower())

# Get the BM25 scores for the word against the document(s) in the corpus
# Since our corpus has one document (the sentence), we get one score.
doc_scores = bm25_model.get_scores(tokenized_query)

print(f"BM25 score for '{word_to_find}' in the sentence: {doc_scores[0]}")
print(f"Note: BM25 provides a relevance score, not a dense vector embedding.")
```

#### 2. Sentence Transformer Embeddings

Sentence Transformers produce dense vector embeddings for sentences, paragraphs, and even words, capturing their semantic meaning. We'll get the embedding for the entire sentence and the individual word to see how they are represented in a vector space.

```python
# Load a pre-trained sentence transformer model
# 'all-MiniLM-L6-v2' is a good balance of speed and performance
model = SentenceTransformer('all-MiniLM-L6-v2')

# Get the embedding for the entire sentence
sentence_embedding = model.encode(sentence, convert_to_tensor=True)

# Get the embedding for the individual word
word_embedding = model.encode(word_to_find, convert_to_tensor=True)

print(f"Sentence Embedding (shape: {sentence_embedding.shape}):\n{sentence_embedding[:5].tolist()}...") # Display first 5 elements
print(f"\nWord Embedding (shape: {word_embedding.shape}):\n{word_embedding[:5].tolist()}...") # Display first 5 elements

print("\nThese are dense vector representations of the sentence and the word, capturing their semantic meaning. You can use these vectors for tasks like similarity comparisons.")
```
